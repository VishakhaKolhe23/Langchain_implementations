```python
from langchain_core.prompts import ChatPromptTemplate
from langchain_ollama import ChatOllama

llm = ChatOllama(model = "gpt-oss:120b-cloud")

prompt = ChatPromptTemplate.from_messages([
    ("system","you are a helpful chatbot"),
    ("human","chat with me and answer my following questions:{topic}"),
])

chat_history = []

while True:
    user_input = input("Enter a question or topic you want to chat about (or type 'exit' to quit): ")
    if user_input.lower() == 'exit':
        break

    messages = prompt.format_messages(topic=user_input)
    response = llm.invoke(messages)
    chat_history.append((messages, response.content))
    print(response.content)
```

```python
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.messages import HumanMessage, AIMessage
from langchain_ollama import ChatOllama

llm = ChatOllama(model="gpt-oss:120b-cloud")

chat_history = []

while True:
    user_input = input("You: ")

    if user_input.lower() == "exit":
        break

    messages = [
        ("system", "You are a helpful chatbot.")
    ]

    # Add old conversation
    for human, ai in chat_history:
        messages.append(("human", human))
        messages.append(("ai", ai))

    # Add current message
    messages.append(("human", user_input))

    prompt = ChatPromptTemplate.from_messages(messages)

    response = llm.invoke(prompt.format_messages())

    print("Bot:", response.content)

    # Save conversation
    chat_history.append((user_input, response.content))
```
