```python
import tiktoken

prompt = "Generative AI (GenAI) is a branch of artificial intelligence focused on creating new content—such as text, images, audio, video, or code—by learning patterns from vast datasets and generating outputs that resemble human-created work."

encoding = tiktoken.get_encoding("cl100k_base")

tokens = encoding.encode(prompt)

print(tokens)

for token in tokens:
    print(f"{encoding.decode([token])} - {token}")
```
