```python

from langchain_core.prompts import ChatPromptTemplate, MessagesPlaceholder
from langchain_ollama import ChatOllama
```

```python


memory = ConversationBufferMemory(return_messages=True)
memory.save_context(
    {"input": "My name is Vishakha"},
    {"output": "Nice to meet you Vishakha"}
)
chat_prompt_template = ChatPromptTemplate.from_messages([
    ("system", "You are a helpful assistant"),
    MessagesPlaceholder(variable_name="chat_history"),
    ("human", "What is my name?")
])

prompt = chat_prompt_template.invoke({
    "chat_history": memory.load_memory_variables({})["history"]
})


response = model.invoke(prompt)

print(response.content)
```

ConversationSummary 

```python
from langchain_classic.memory import ConversationBufferMemory, ConversationBufferWindowMemory,  ConversationSummaryMemory , ConversationSummaryBufferMemory
from langchain_classic.chains import ConversationChain
from langchain_ollama import ChatOllama
memory=ConversationBufferMemory()
llm = ChatOllama(model="gpt-oss:120b-cloud")
```

```python
summary=ConversationSummaryMemory(llm=llm, memory_key="history")
conversation_chain=ConversationChain(llm=llm,memory=summary)
response_1= conversation_chain.invoke("My name is Vishakha")
response_2= conversation_chain.invoke("I work with Zensar")
response_3= conversation_chain.invoke("What is my name? ")
final_summary=summary.load_memory_variables({})["history"]
print(type(final_summary))
print(final_summary)
```

```python
summary=ConversationSummaryBufferMemory(llm=llm , memory_key="history", max_token_limit=10)
conversation_chain=ConversationChain(llm=llm , memory=summary)
response_1= conversation_chain.invoke("My name is Vishakha")
response_2= conversation_chain.invoke("I work with Zensar")
response_3= conversation_chain.invoke("What is my name? ")
final_summary=summary.load_memory_variables({})["history"]
print("\nFinal Conversation Summary:")
print(final_summary)
```

```python
documents=["I like music",
           "I like cricket", 
           "I like swimming"]

vector_store = FAISS.from_texts(documents, embedding_model)

query = "What are my hobbies?"
result = vector_store.similarity_search(query)

for r in result:
    print(r.page_content)

```
